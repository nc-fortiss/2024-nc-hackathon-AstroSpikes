```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_float = pd.read_csv("Float/kpts_predictions_pose_metrics.csv")
df_quant = pd.read_csv("Quantized/kpts_predictions_pose_metrics.csv")
```

```python
import matplotlib.ticker as mticker
from matplotlib.ticker import LogLocator

scores_float = df_float["speed_score"].astype(np.float32).dropna().to_numpy()
scores_quant = df_quant["speed_score"].astype(np.float32).dropna().to_numpy()
scores_float.sort()
scores_quant.sort()

n_float = len(scores_float)
n_quant = len(scores_quant)
y_float = np.arange(1, n_float + 1) / n_float
y_quant = np.arange(1, n_quant + 1) / n_quant

eps = 0.000029   # pick something smaller than your step 1/n
y_plot_float = np.clip(y_float, eps, 1 - eps)
y_plot_quant = np.clip(y_quant, eps, 1 - eps)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(scores_float, y_plot_float, linestyle="-", marker="o", markersize=5, label="Float")
ax.plot(scores_quant, y_plot_quant, linestyle="-", marker="o", markersize=5, label="Quantized")

# x axis formatting
ax.set_xlabel("SPEED Score")

# y axis formatting
ax.set_ylabel("Fraction of Test Set ($\leq s$)")


ax.set_yscale('logit')
ax.set_ylim(0, 1)

# Manual ticks/labels (adjust as you like; keep within [eps, 1-eps])
ticks = [0.003, 0.005, 0.01, 0.02, 0.05, 0.1, 0.5, 0.8, 0.9, 0.95, 0.98, 0.99, 0.995, 0.997]
ticks = [t for t in ticks if eps < t < 1-eps]
ax.yaxis.set_major_locator(mticker.FixedLocator(ticks))
ax.yaxis.set_major_formatter(mticker.FixedFormatter([f'{t:g}' for t in ticks]))
ax.set_xlim(0,5)

# ax.set_ylim(max(1/n, 3e-3), 0.997)
# ax.set_xlim(0, 10)
ax.grid(True, which='both', alpha=0.3)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
def compute_avgs(df):
    loc_err_avg = df["loc_err"].mean()
    ori_err_avg = df["ori_err"].mean()
    speed_score_avg = df["speed_score"].mean()
    return {
        "loc_err_avg": loc_err_avg,
        "ori_err_avg": ori_err_avg,
        "speed_score_avg": speed_score_avg
    }

compute_avgs_float = compute_avgs(df_float)
compute_avgs_quant = compute_avgs(df_quant)
print("Float Averages:")
print(compute_avgs_float)
print("\nQuantized Averages:")
print(compute_avgs_quant)
```

```python
# Define pose error metric (example: you can replace this)
# If you only have loc_err and ori_err, combine them or pick one
df_float["pose_err"] = df_float["norm_loc_err"] + df_float["ori_err"]  # or a combination
df_quant["pose_err"] = df_quant["norm_loc_err"] + df_quant["ori_err"]  # or a combination

# Sort errors
err_float_sorted = np.sort(df_float["pose_err"].values)
err_quant_sorted = np.sort(df_quant["pose_err"].values)

# Generate ranks
rank_float = np.arange(1, len(err_float_sorted)+1)
rank_quant = np.arange(1, len(err_quant_sorted)+1)

# Plot
plt.figure(figsize=(8,5))
plt.plot(rank_float, err_float_sorted, 'o-', color='black', label="Float Model")
plt.plot(rank_quant, err_quant_sorted, 'o-', color='red', label="Quantized Model")

# Example baseline (mean error for each model)
plt.axhline(err_float_sorted.mean(), color='black', linestyle='--', label="Float Mean")
plt.axhline(err_quant_sorted.mean(), color='red', linestyle='--', label="Quant Mean")

# Formatting
plt.yscale("log")  # Log scale like in your example
plt.xlabel("Rank")
plt.ylabel("Pose error")
plt.title("Ranked Pose Error Comparison")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
